```python
from roboflow import Roboflow
from ultralytics import YOLO
import os
```

## 2. Download dataset from Roboflow

```python
API_KEY = "YOUR_API_KEY"  # TODO: replace with your Roboflow API key
WORKSPACE = "workspace-5ujvu"
PROJECT_NAME = "basketball-players-fy4c2-vfsuv"
VERSION = 17
DATASET_FORMAT = "yolov5"  # or "yolov8"

rf = Roboflow(api_key=API_KEY)
project = rf.workspace(WORKSPACE).project(PROJECT_NAME)
version = project.version(VERSION)
dataset = version.download(DATASET_FORMAT)

dataset_location = dataset.location
print(f"Dataset downloaded to: {dataset_location}")
```

## 4. Train YOLO model

```python
MODEL_PATH = "yolov5l6u.pt"  # base weights to fine-tune from
DATA_YAML_PATH = os.path.join(dataset_location, "data.yaml")
IMG_SIZE = 640
EPOCHS = 100
BATCH = 8
```

```python
model = YOLO(MODEL_PATH)

results = model.train(
    data=DATA_YAML_PATH,
    epochs=EPOCHS,
    imgsz=IMG_SIZE,
    batch=BATCH,
    plots=True,
)
```
